Q1. Define Covariance and explain how it differs from Correlation in terms of scale and interpretation.

Covariance

Covariance measures the direction in which two variables change together.

Positive covariance → both variables tend to increase together.
Negative covariance → one increases while the other decreases.
Its value depends on the scale/units of the variables, so it is difficult to compare across different datasets.

Correlation is the standardized form of covariance. It always lies between -1 and +1, so it is easier to interpret and compare.

Q2. What does a positive, negative, and zero covariance indicate about the relationship between two variables?

Positive covariance: Both variables tend to increase or decrease together.
Negative covariance: When one variable increases, the other tends to decrease.
Zero covariance: There is no linear relationship between the two variables.

Example:
Height and weight → Positive covariance
Price and demand → Negative covariance
Two unrelated variables → Zero covariance.

Q3. Discuss the limitations of covariance as a measure of relationship between two variables. Why is correlation
preferred in many cases?

Covariance tells you the direction of a linear relationship between two variables, but it has key limitations:

Dependent on units: Covariance is bound to the units of measurement of the variables. For example, measuring height in meters versus centimeters drastically changes the covariance value, even though the actual relationship stays identical.

No standard scale: Its value ranges from negative infinity to positive infinity. Because there is no upper or lower bound, you cannot tell the strength of the relationship just by looking at the number. A covariance of 50 isn't necessarily stronger than a covariance of 5.

Why Correlation is Preferred
Correlation addresses these limitations by normalizing covariance:

Scale-invariant: It scales the value to a fixed range between -1 and +1, making it unitless.

Easy to interpret: The magnitude directly indicates strength — values close to +1 or -1 show a strong linear relationship, while values near 0 show little to no linear relationship.

Direct comparisons: Because it is unitless, you can easily compare relationships across entirely different pairs of variables (e.g., height vs. weight compared to income vs. age).

Q4. Explain the difference between Pearson’s correlation coefficient and Spearman’s rank correlation
coefficient. When would you prefer to use Spearman’s correlation?

Pearson’s correlation measures linear relationships between continuous data, assuming normal distribution and high sensitivity to outliers.

Spearman’s rank correlation measures monotonic relationships by converting data into ranks, requiring no normal distribution and offering resistance to outliers.

Key Differences
Feature	Pearson	Spearman
Relationship	Strictly linear	Monotonic (curves allowed)
Data Type	Continuous	Ordinal, ranked, or continuous
Sensitivity	Sensitive to outliers	Robust to outliers
Assumption	Normal distribution	Non-parametric (any distribution)
When to Prefer Spearman
Ordinal/Ranked Data: Customer ratings, survey scores, or competition ranks.

Non-Linear Relationships: Variables move together consistently, but not in a straight line.

Outliers Present: Data contains extreme values that would skew Pearson's result.

Q5. If the correlation coefficient between two variables X and Y is 0.85, interpret this value in context. Can you
infer causation from this value? Why or why not?

A correlation coefficient ($r$) of 0.85 indicates a strong, positive linear relationship between variables $X$ and $Y$. As the values of $X$ increase, the values of $Y$ also tend to increase consistently.Inferring CausationNo, you cannot infer causation from this value alone. Correlation simply measures how two variables move together; it does not prove that $X$ causes $Y$ or vice versa.Why Causation Cannot Be InferredConfounding/Lurking Variables: A third, unmeasured variable ($Z$) might be influencing both $X$ and $Y$ simultaneously (e.g., ice cream sales and sunburn rates both correlate with summer temperatures).Reverse Causality: $Y$ could actually be causing $X$ instead of $X$ causing $Y$.Spurious Correlation: High correlations can happen entirely by coincidence, especially in observational data without controlled experimentation.To establish causation, you need controlled experiments, longitudinal studies, and a clear theoretical mechanism — not just a high statistical correlation.
